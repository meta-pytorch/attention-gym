# A tour of warpdrive

warpdrive is a thin CuTeDSL layer for warp-specialized Blackwell kernels. Each role (load,
MMA, epilogue, ...) is a method. Declared channels connect the roles. The layer derives:

- barrier counts and phases, and TMA byte arming;
- role dispatch, `setmaxnreg` and warpgroup padding;
- TMEM allocation and the cross-CTA plumbing.

It also checks the protocol while tracing. Tiles, layouts, stage counts, warps and the math
stay yours, in ordinary CuTeDSL.

The tour starts with a two-warp producer/consumer kernel and builds to a pipelined
TMA → UMMA → epilogue GEMM. Then it covers scheduling, clusters, storage aliasing, the checkers
and `describe()`. It ends by mapping `persistent_gemm.py` and `fa4_fwd.py` onto the nouns.

| § | Nouns | Verbs |
|---|---|---|
| 1 | `WarpSpecialized`, `@role`, `RoleConfig`, `Ring`, `Threads`/`Warps`, `Producer`/`Consumer`, `Handle`, `Token`, `Preview`, `Slots`, `Sharing`, `NamedBarrier`, `Turnstile` | `acquire`/`commit`/`produce`, `wait`/`release`/`consume`, `try_wait`, `peek`, `describe()` |
| 2 | `Tma`, `Umma`, `SmemPlan`, `StageView`, `TmemPlan`, `Tmem` | `h.view`, `h.barrier`, `tmem.ptr` |
| 3 | `WorkStream`, `WorkMap`, `Raster`, schedules, `WorkCursor` | `work.item`, `work.advance()`, `bind`, `launch_args` |
| 4 | `CtaPair`, `Cluster`, `cta="leader"`, multicast `Tma` | `cluster.multicast` |
| 5 | `overlaps=`, `TmemPlan.alias`, access annotations | `alias_check` |
| 6 | `ProtocolError`, protocol trace | `trace_protocol`, `unchecked` |
| 7 | `raw=True` roles, `external=`, `override=`, `Barriers` | `signal=False`, `cute.arch` |
| 8 | `persistent_gemm.py`, `fa4_fwd.py` | |

## 1. The smallest kernel: two roles and a ring

A `WarpSpecialized` subclass declares its channels as attributes in `__init__`. Its roles are
methods decorated `@role(warps=...)` directly above `@cute.jit`. `warps=0` means warp index 0,
not zero warps.

Role parameters bind like this:

- A parameter annotated `Producer` or `Consumer` receives that end of the channel with the
  same name: `ch: Producer` binds to `self.ch`.
- The single unannotated parameter receives the context passed to `run_roles`.

Every role returns every state argument it took: channel ends, work cursors, turnstile seats.

`Ring(2, Threads, Threads, payload=...)` reads: two reusable stages; every producer thread
arrives to publish a stage; every consumer thread arrives to hand it back. A **stage** is a
reusable storage slot. An **epoch** is one exchanged value; epochs cycle through the stages.
The payload is SMEM the layer owns, of shape `(32, stages)`. `h.data` is the handle's stage.

The producer below spells out the verbs: `acquire`, write, `commit`. The consumer uses the
context-manager form: `consume()` waits on entry and releases on exit. `style` swaps in the
other consumer verbs used later in this section.

````python
WORDS = 32  # one Int32 per lane per stage


class Relay(WarpSpecialized):
    """Warp 0 writes epoch e's words ``e * 32 + lane``; the consumer sums them per lane."""

    def __init__(self, stages: int = 2, release=Threads, style: str = "with", roles=None):
        self.style = style
        self.roles = roles  # {role name: RoleConfig} over the @role defaults
        self.ch = Ring(stages, Threads, release, payload=(Int32, (WORDS,)))

    @role(warps=0)
    @cute.jit
    def producer(self, ctx, *, ch: Producer):
        lane = cute.arch.lane_idx()
        for e in cutlass.range(ctx.n, unroll=1):
            h = ch.acquire()  # wait until stage h.index is empty; the end moves on
            h.data[lane] = e * WORDS + lane
            h.commit()  # publish: every producer thread arrives on FULL
        return ch

    @role(warps=1)
    @cute.jit
    def consumer(self, ctx, *, ch: Consumer):
        lane = cute.arch.lane_idx()
        total = Int32(0)
        for _ in cutlass.range(ctx.n, unroll=1):
            if cutlass.const_expr(self.style == "token"):
                token = ch.try_wait()  # one probe of FULL; no handle, no advance
                h = ch.wait(token=token)  # blocks only if the stage was not yet full
                total += h.data[lane]
                h.release()
            elif cutlass.const_expr(self.style == "peek"):
                preview = ch.peek()  # waits for FULL, read-only, does not advance
                total += preview.view()[lane]
                h = ch.wait()  # owns the previewed epoch; its release covers the read
                h.release()
            else:
                with ch.consume() as h:
                    total += h.data[lane]
        ctx.out[lane] = total  # every consumer warp writes the same sums
        return ch

    @cute.kernel
    def kernel(self, out: cute.Tensor, n: Int32):
        self.begin_roles()  # allocate + initialize barriers and payloads
        self.run_roles(SimpleNamespace(out=out, n=n))  # fence, sync, dispatch, teardown

    @cute.jit
    def __call__(self, out: cute.Tensor, n: Int32, stream):
        self.kernel(out, n).launch(**self.launch_config(grid=(1, 1, 1)), stream=stream)


print(Relay().describe())
````

`describe()` runs before anything is compiled. It collects the declarations and prints what
the layer derived. `full 32 x2` means each of the two FULL barriers expects 32 arrivals: the
producer warp's threads. `empty 32 x2` counts the consumer's 32.

Next, compile with fake compact tensors, launch, and check against a host reference. Dynamic
sizes stay runtime values. `n` is a runtime epoch count: 1, 2 and 7 exercise the first stage,

**Caching.** The library's examples compile through `jit_cache`. It keeps kernels on disk,

````python
def build_relay(op: Relay):
    out = make_fake_compact_tensor(Int32, (WORDS,), assumed_align=4)
    return compile_tvm_ffi(op, out, Int32(1), name="tour_relay")


def check_relay(fn, n: int) -> None:
    out = torch.full((WORDS,), -1, device="cuda", dtype=torch.int32)
    fn(out, n)
    lane = torch.arange(WORDS, device="cuda", dtype=torch.int32)
    torch.testing.assert_close(out, WORDS * n * (n - 1) // 2 + n * lane)


relay = build_relay(Relay())
for n in (1, 2, 7, 1000):
    check_relay(relay, n)
print("Relay matches the reference for n = 1, 2, 7, 1000")
````

### Mechanisms and placement change the counts, not the role bodies

- `Threads` counts every thread of the bound roles.
- `Warps`: every lane runs the side's fence and `bar.warp.sync`, then one elected lane per
  warp arrives.
- `RoleConfig` replaces a role's `@role` placement from the constructor (`self.roles`).
  Moving the consumer to warps 4–7 makes it a four-warp role.
- A `regs=` budget is per warpgroup (`setmaxnreg`). Roles that set one pad the launch to
  whole warpgroups, and the padding warps take their group's budget.

The consumer's other verbs:

- `try_wait()` probes FULL once and returns a `Token`. `wait(token=...)` skips the blocking
  wait when the probe succeeded.
- `peek()` waits for the next epoch and returns a read-only `Preview` without advancing. The
  next `wait()` owns that epoch.

Tracing checks both: a stale token, or a preview with no owning wait, is a `ProtocolError`.

````python
wide = Relay(release=Warps, roles={"consumer": RoleConfig(range(4, 8))})
show(wide, "Relay", "Roles", "Channels")
budgets = Relay(
    roles={"producer": RoleConfig(0, regs=40), "consumer": RoleConfig(range(4, 8), regs=232)}
)
print()
show(budgets, "Roles")
print()

for name, op in [
    ("four-warp Warps consumer", wide),
    ("try_wait token", Relay(style="token")),
    ("peek preview", Relay(style="peek")),
]:
    fn = build_relay(op)
    for n in (1, 7, 1000):
        check_relay(fn, n)
    print(f"{name}: matches")

print("\nWarps release, elected per warp after a warp sync:")
fake_out = make_fake_compact_tensor(Int32, (WORDS,), assumed_align=4)
wide_ptx = compile_with_ptx(
    Relay(release=Warps, roles={"consumer": RoleConfig(range(4, 8))}),
    fake_out,
    Int32(1),
    name="relay_warps",
).__ptx__
sync_ops(wide_ptx)
````

With `Warps`, the EMPTY count drops from 128 (four warps × 32 threads) to 4. The release
gains a `bar.warp.sync` and an `elect.sync` before its arrive. The FULL side is unchanged.

With budgets, warps 1–3 run no role but share warp 0's warpgroup, so they take its 40
registers. `setmaxnreg` only moves registers, so `describe()` rejects a split that does not
fit the registers the launch starts with.

### The rest of the channel vocabulary

A declaration-only kernel, whose role bodies only return their state, is enough to see what
the layer derives for the other channel shapes. `describe()` needs no compile.

| Declaration | Meaning |
|---|---|
| `Slots(n, ...)` | The role picks the slot (`s.wait(j)`). `Consumer[Instance]` binds each instance of a multi-instance role to its own slot: `softmax[0]@0` in the output. |
| `Sharing.ROUND_ROBIN` | Instances take a side's epochs in turn. Counts are one instance's. |
| `Sharing.EXCLUSIVE` | Each epoch goes to one of several consumer roles, which `skip()` the rest. |
| `sync="named"` | A Threads → Threads channel on CTA named barriers instead of mbarriers. |
| `NamedBarrier(threads=N)` | One allocated barrier ID, for a rendezvous or a one-shot flag. |
| `Turnstile` | A round-robin critical section across a role's instances (`with mufu.turn():`). |

````python
class Vocabulary(WarpSpecialized):
    def __init__(self):
        self.s = Slots(2, Umma, Threads)  # MMA fills slot j, softmax[j] reads it
        self.rr = Ring(4, Threads, Threads, consumer_sharing=Sharing.ROUND_ROBIN)
        self.row_max = Ring(1, Threads, Threads, sync="named")
        self.done = NamedBarrier(threads=256 + 32)
        self.mufu = Turnstile()

    @role(warps=8)
    @cute.jit
    def mma(self, ctx, *, s: Producer, rr: Producer, row_max: Producer):
        return s, rr, row_max

    @role(warps=[range(4), range(4, 8)])
    @cute.jit
    def softmax(
        self,
        ctx,
        *,
        stage: Instance,
        s: Consumer[Instance],
        rr: Consumer,
        row_max: Consumer,
        mufu: Turn,
    ):
        return s, rr, row_max, mufu


show(Vocabulary(), "Roles", "Channels", "Named", "Turnstiles")
````

## 2. A pipelined GEMM: TMA → UMMA → epilogue

[`attn_gym/warpdrive/examples/gemm.py`](https://github.com/meta-pytorch/attention-gym/blob/main/attn_gym/warpdrive/examples/gemm.py) computes BF16 `D = A @ B.T`, one 128 × 256 tile per
CTA:

- warp 4 loads A/B with TMA;
- warp 5 issues `tcgen05` MMAs;
- warps 0–3 read the accumulator from TMEM and store D.

Two channels carry the handoffs:

| Channel | Declaration | Producer → consumer | Epochs per tile |
|---|---|---|---|
| `ab` | `Ring(3, Tma(49152), Umma, buffer={"A": a, "B": b})` | TMA bytes complete FULL → MMA completion (`tcgen05.commit`) frees the stage | `ceil(K / 64)` |
| `acc` | `Ring(1, Umma, Threads)` | MMA completion publishes → 128 epilogue threads release | 1 |

New nouns:

- `Tma(bytes)`: `acquire` arms the FULL barrier with the stage's byte count. The copies,
  passed `tma_bar_ptr=h.barrier`, complete it.
- `Umma`: an elected `tcgen05.commit` signals when the issuing thread's MMAs finish. The stage
  is reused only after the MMA has read it.
- `SmemPlan` declares named buffers. `buffer=` binds them to a channel, and `begin_roles`
  allocates them from the layouts you pass.
- `StageView(tensor, axis)` registers the stage mode of a TMA or MMA partition. `h.view("A")`
  selects the handle's stage of it.
- `TmemPlan` names TMEM column ranges. A role taking `tmem: Tmem` gets `tmem.ptr("acc")` and
  joins the coordinated free.

The MMA role shows how the handles nest. It holds `acc.produce()` around the whole K loop and
opens one `ab.consume()` per K tile:

````python
from attn_gym.warpdrive.examples.gemm import Gemm, gemm

print(inspect.getsource(Gemm.mma))
print(Gemm().describe())
````

`tx 49152 B` is `(128 + 256) × 64 × 2` bytes per stage. Tracing fills in
`tma not checked yet` and the missing SMEM sizes: the layouts bind in `begin_roles`, and the
TMA check runs on the traced load role (next cells). The 160/40/40 budgets pad the launch to
two warpgroups, so warps 6–7 run nothing.

Next, check correctness against an FP32 reference, with PyTorch's BF16 error as the
yardstick. Then time it at 8192³.

Timing method: profiler device time, mean over repeats, L2 flushed before each call
(Inductor's `do_bench_using_profiling`). Fixed tensors and default clocks; a microbenchmark.

````python
def gemm_error(fn, m: int, n: int, k: int, **kwargs) -> tuple[float, float]:
    a = torch.randn(m, k, device="cuda", dtype=torch.bfloat16)
    b = torch.randn(n, k, device="cuda", dtype=torch.bfloat16)
    ref = a.float() @ b.float().T
    err = (fn(a, b, **kwargs).float() - ref).abs().max().item()
    return err, ((a @ b.T).float() - ref).abs().max().item()


for shape in [(512, 512, 512), (300, 520, 136), (4097, 2056, 4104)]:
    err, base = gemm_error(gemm, *shape)
    print(f"M, N, K = {shape}: max |err| {err:.3g} (torch bf16 {base:.3g})")
    assert err <= base * 1.01 + 1e-3

a8 = torch.randn(8192, 8192, device="cuda", dtype=torch.bfloat16)
b8 = torch.randn_like(a8)
flops8 = 2 * 8192**3
print("warpdrive gemm ", tflops(flops8, bench_us(gemm, a8, b8)))
print("torch.mm       ", tflops(flops8, bench_us(torch.mm, a8, b8.T)))
````

This GEMM is the teaching kernel: one tile per CTA, and no overlap between epilogue and
mainloop. The persistent GEMM in §4 closes most of the gap to cuBLAS.

## 3. Scheduling: which work each CTA runs

`self.work = WorkStream(mapping=..., schedule=...)` gives every role of a work unit the same
sequence of items. A **work unit** is a CTA, or a whole cluster.

- The **schedule** picks the linear indices for each unit.
- The **mapping** (`WorkMap`) decodes an index into an item.

A role takes `work: WorkCursor` and drains it with exactly this loop:

```python
while work.valid:
    m, n, _ = work.item  # read before advancing
    ...
    work.advance()  # once, at the end of the body
return ..., work
```

The toy below records which CTA ran each item, and in what order. `LowerTriangle` is a custom
`WorkMap`. It decodes a raster index and drops tiles above the diagonal (`skips = True`), as a
causal mask would.

````python
@dataclass(frozen=True)
class LowerTriangle(WorkMap):
    """Raster over ``(m_tiles, n_tiles, 1)``, m fastest, keeping tiles with ``n <= m``."""

    words = 3
    skips = True  # decode may drop an index for every role

    def count64(self, space):
        m_tiles, n_tiles, _ = space
        return Int64(m_tiles) * n_tiles

    def decode(self, index, space, state):
        m_tiles = space[0]
        m, n = index % m_tiles, index // m_tiles
        return n <= m, (m, n, 0), state  # (keep, words, next state)

    def get_key(self) -> str:
        return "lower_triangle"


class WhoRunsWhat(WarpSpecialized):
    def __init__(self, schedule, mapping=None):
        self.work = WorkStream(mapping=mapping or Raster(), schedule=schedule)

    @role(warps=0)
    @cute.jit
    def worker(self, ctx, *, work: WorkCursor):
        bx, by, _ = cute.arch.block_idx()
        gx, _, _ = cute.arch.grid_dim()
        cta, step = bx + by * gx, Int32(0)  # linear CTA id (OneShot launches a 2-D grid)
        while work.valid:
            m, n, _ = work.item
            if cute.arch.lane_idx() == 0:
                ctx.owner[m, n] = cta
                ctx.order[m, n] = step
            step += 1
            work.advance()
        return work

    @cute.kernel
    def kernel(self, owner: cute.Tensor, order: cute.Tensor, bound):
        self.begin_roles(work=bound)
        self.run_roles(SimpleNamespace(owner=owner, order=order))

    @cute.jit
    def __call__(
        self,
        owner: cute.Tensor,
        order: cute.Tensor,
        workspace: cute.Tensor,
        max_ctas: Int32,
        stream,
    ):
        bound = self.work.bind((owner.shape[0], owner.shape[1], 1), workspace, max_ctas)
        self.kernel(owner, order, bound).launch(**self.launch_config(bound), stream=stream)


def who_runs_what(schedule, mapping=None, shape=(4, 6), max_ctas=None):
    op = WhoRunsWhat(schedule, mapping)
    grid = make_fake_compact_tensor(Int32, (cute.sym_int(), cute.sym_int()), assumed_align=4)
    workspace = make_fake_compact_tensor(Int32, (cute.sym_int(),), assumed_align=4)
    fn = compile_tvm_ffi(op, grid, grid, workspace, Int32(1), name="tour_who_runs_what")
    owner = torch.full(shape, -1, device="cuda", dtype=torch.int32)
    order = torch.full_like(owner, -1)
    # AtomicPersistent's zeroed counter (a placeholder otherwise) and the resident-CTA count
    workspace, sms = op.work.launch_args(owner.device)
    fn(owner, order, workspace, max_ctas or sms)
    return op, owner.cpu(), order.cpu()


for label, schedule, mapping in [
    ("OneShot(), Raster()", OneShot(), None),
    ("StaticPersistent(), Raster()", StaticPersistent(), None),
    ("StaticPersistent(), Raster(swizzle=2)", StaticPersistent(), Raster(swizzle=2)),
    ("StaticPersistent(), LowerTriangle()", StaticPersistent(), LowerTriangle()),
]:
    _, owner, order = who_runs_what(schedule, mapping, max_ctas=3)
    print(f"{label}: CTA of each (m, n), then its visit order")
    for owner_row, order_row in zip(owner.tolist(), order.tolist()):
        print("  ", " ".join(f"{c:2d}" for c in owner_row), "   ", order_row)
````

- `OneShot` launches the tile space, one CTA per item.
- `StaticPersistent` with three CTAs strides the indices by 3.
- Swizzling changes which item an index decodes to, not the strides.
- `LowerTriangle` drops items from every role's sequence.

The broadcast schedules add a scheduler warp. It publishes each item to the roles over a
generated `work_ring` channel. On 65536 items:

- `AtomicPersistent` claims work from a global counter, the `workspace` from `launch_args`.
  It launches `max_ctas` CTAs (the SM count `launch_args` estimates), which loop over the
  counter.
- `ClusterLaunchControl` (CLC) cancels CTAs that have not launched yet and takes their work.
  It launches the whole grid. Each resident CTA keeps cancelling unlaunched CTAs and runs
  their items, so about as many CTAs run as fit at once. This tiny kernel fits many per SM.

The counts are a runtime snapshot. `work_clc` is CLC's 16-byte cancel-response slot, filled by
`clusterlaunchcontrol.try_cancel`:

````python
for schedule in (AtomicPersistent(), ClusterLaunchControl()):
    op, owner, _ = who_runs_what(schedule, shape=(256, 256))
    ran = torch.bincount(owner.flatten())
    print(
        f"{schedule.describe()}: {(ran > 0).sum().item()} CTAs ran the {owner.numel()} items, "
        f"at most {ran.max().item()} each"
    )
show(op, "Roles", "Channels", "Streams")
````

## 4. Clusters and CTA pairs

A cluster owns one work item. `self.cluster = CtaPair()` runs 2-CTA (`CtaGroup.TWO`) MMAs:

- each CTA loads half the operands;
- the leader's MMA warp (`@role(..., cta="leader")`) issues MMAs that read both CTAs' SMEM;
- each epilogue stores its half from its own TMEM.

The declaration derives which barriers every signal reaches. It does not convert atoms or
layouts: you double the MMA tile's M and build atoms with `self.cluster.cta_group`.

`Cluster(2, 2, pair=True)` adds a second pair along N. Both pairs need the same rows of A,
so each CTA loads part of the A tile and multicasts it: `Tma(bytes, multicast="n")`, with
masks from `self.cluster.multicast("n")`.

[`attn_gym/warpdrive/examples/persistent_gemm.py`](https://github.com/meta-pytorch/attention-gym/blob/main/attn_gym/warpdrive/examples/persistent_gemm.py) takes the schedule and the
cluster as constructor arguments and keeps one set of role bodies. It runs `gemm.py`'s roles
in a work loop, plus:

- two TMEM accumulator stages, so the MMA starts tile i+1 while the epilogue drains tile i;
- a TMA-store epilogue.

Compare the derived `ab` and `acc` channels:

````python
from attn_gym.warpdrive.examples.persistent_gemm import (
    PersistentGemm,
    compile_persistent_gemm,
    persistent_gemm,
)

for cluster in (CtaPair(), Cluster(2, 2, pair=True)):
    op = PersistentGemm(cluster=cluster)
    compile_persistent_gemm(op)  # fills the TMA check line
    print(f"{cluster}:")
    show_channels(op, "ab", "acc")
    print()
````

Reading the pair:

| Field | Meaning |
|---|---|
| `tx 61440 B (leader-armed)` | The leader's FULL barrier collects both CTAs' bytes, twice the per-CTA 30720. |
| `release_mask 0b11` | The MMA's completion releases A/B in both CTAs. |
| `commit_mask 0b11` | The MMA's completion publishes the accumulator to both CTAs. |
| `empty leader 256 x2` | The leader's EMPTY barrier waits for both epilogues: 2 × 128 threads, which the role code never states. |
| `tail` | The producer drains its last stages before exit, because the remote MMA may still be reading them. |

In the 2 × 2 cluster, the other pair's next multicast overwrites each CTA's stage. So every
CTA's EMPTY waits for both leaders (`2 x6`).

Every schedule and cluster below runs the same role bodies. Errors are on a ragged
4097 × 2056 × 4104 problem against FP32, with PyTorch BF16 for scale. Timings are as in §2,
at 8192³; `torch.mm`'s time is printed there.

````python
configs = [
    ("OneShot, 1 CTA", OneShot(), None),
    ("StaticPersistent, 1 CTA", StaticPersistent(), None),
    ("AtomicPersistent, 1 CTA", AtomicPersistent(), None),
    ("ClusterLaunchControl, 1 CTA", ClusterLaunchControl(), None),
    ("ClusterLaunchControl, CtaPair", ClusterLaunchControl(), CtaPair()),
    ("StaticPersistent, Cluster(2, 2)", StaticPersistent(), Cluster(2, 2, pair=True)),
]
for label, schedule, cluster in configs:
    kwargs = {"schedule": schedule, "cluster": cluster}
    err, base = gemm_error(persistent_gemm, 4097, 2056, 4104, **kwargs)
    assert err <= base * 1.01 + 1e-3, (label, err, base)
    us = bench_us(persistent_gemm, a8, b8, **kwargs)
    print(f"{label:32s} err {err:.3g} (torch {base:.3g})  {tflops(flops8, us)}")
````

## 5. Storage and aliasing

Channels order *when* a stage may be used. `SmemPlan` and `TmemPlan` say *where* it lives.
Overlapping storage is declared, never inferred:

- `smem_plan.buffer("V", ..., overlaps=k)` gives K's bytes a second layout (FA4's V reuses
  K's stages; both bind to one ring).
- `TmemPlan().region("S", ...).alias("P", of="S", ...)` places P inside S's columns.

Overlapping bytes does not order accesses. The alias diagnostic checks the accesses you
annotate at each operation against the channel paths between them. An annotation looks like
`h.view("K", write=True, by="tma")` or `tmem.access("P", j, read=True, by="umma")`. The
diagnostic is a bug finder: `no hazard found` is not a proof.

FA4 forward declares both kinds of alias:

````python
from attn_gym.warpdrive.examples.fa4_fwd import (
    Fa4Params,
    FlashAttentionForward,
    compile_fa4_fwd,
    fa4_fwd,
)

fa4 = FlashAttentionForward()
compile_fa4_fwd(fa4)
show(fa4, "TMEM", "SMEM", "Alias")

b_, s_, h_, d_ = 4, 4096, 16, 128
q, k, v = (torch.randn(b_, s_, h_, d_, device="cuda", dtype=torch.bfloat16) for _ in range(3))
out, lse = fa4_fwd(q, k, v)
ref = F.scaled_dot_product_attention(*(x.transpose(1, 2).float() for x in (q, k, v)))
sdpa = lambda: F.scaled_dot_product_attention(*(x.transpose(1, 2) for x in (q, k, v)))
err = (out.float() - ref.transpose(1, 2)).abs().max().item()
base = (sdpa().float() - ref).abs().max().item()
print(f"\nfa4_fwd max |err| {err:.3g} (SDPA bf16 {base:.3g})")
assert err <= 2 * base
flops = 4 * b_ * h_ * s_ * s_ * d_
print("warpdrive fa4_fwd        ", tflops(flops, bench_us(fa4_fwd, q, k, v)))
print("warpdrive fa4_fwd, pair  ", tflops(flops, bench_us(fa4_fwd, q, k, v, cluster=CtaPair())))
print("torch SDPA               ", tflops(flops, bench_us(sdpa)))
````

### A hazard the diagnostic finds

`overlap_o_q=True` places the output staging buffer O over Q (FA4's `overlap_sO_sQ`). With
one item per CTA that is safe. Under a persistent schedule, the load warp may land the next
item's Q while the epilogue's TMA store still reads O from the same bytes, unless something
orders them.

`OverlapOQ(ordered=True)` adds that edge as a one-stage, payload-free `q_free` channel from
epilogue to load. `ordered=False` drops it. Both compile; only the diagnostic tells them
apart.

About the setup:

- `trace_protocol = True` also keeps these notebook-defined subclasses out of the disk cache.
  The example's compile helper builds traced kernels uncached.
- The example's constructor rejects `overlap_o_q` with a persistent schedule. So `OverlapOQ`
  swaps in the stream after construction and reuses the example's helpers `load_item` and
  `epilogue_item`.
- Neither variant is launched.

````python
class OverlapOQ(FlashAttentionForward):
    def __init__(self, ordered: bool):
        super().__init__(overlap_o_q=True)
        self.ordered = ordered
        self.q_free = Ring(1, Threads, Threads)  # epilogue: "O stored"; load: next Q may land
        self.work = WorkStream(mapping=Raster(along="m"), schedule=StaticPersistent())
        self.trace_protocol = True

    @role(warps=14, regs=48)
    @cute.jit
    def load(
        self, params: Fa4Params, *, q: Producer, kv: Producer, q_free: Consumer, work: WorkCursor
    ):
        rank = self.cluster.rank()
        thr = params.mma_qk.get_slice(rank), params.mma_pv.get_slice(rank)
        while work.valid:
            q, kv = self.load_item(params, q, kv, thr, work.item)
            if cutlass.const_expr(self.ordered):
                with q_free.consume():
                    pass
            work.advance()
        return q, kv, q_free, work

    @role(warps=13, regs=48)
    @cute.jit
    def epilogue(self, params: Fa4Params, *, epi: Consumer, q_free: Producer, work: WorkCursor):
        rank = self.cluster.rank()
        while work.valid:
            epi = self.epilogue_item(params, epi, rank, work.item)
            if cutlass.const_expr(self.ordered):
                with q_free.produce():
                    pass
            work.advance()
        return epi, q_free, work


for ordered in (False, True):
    op = OverlapOQ(ordered)
    with warnings.catch_warnings(record=True) as caught:
        warnings.simplefilter("always")
        compile_fa4_fwd(op)
    print(f"ordered={ordered}: {len(caught)} warning(s)", *(w.message for w in caught))
    text = op.describe()
    in_pair = False
    for line in text[text.index("Alias diagnostic") :].split("\n\n")[0].splitlines()[1:]:
        if not line.startswith("    "):  # a pair's first line; its findings are indented
            in_pair = "O ~ Q" in line
        if in_pair:
            print(line)
    print()
````

Each `[channel]` finding names both accesses, their source lines and work items, and the
missing edge (`no channel path orders them`). The correction warps' O stores and the
epilogue's TMA read of O both race the next item's Q load.

The `[proxy]` lines are the weaker case. The only path from those generic-proxy O stores to
the MMA's next read of Q runs through `s`, whose release fences TMEM stores only. With
`q_free`, the path runs through `epi`, whose commit has the `smem_async` fence, and every
finding goes away.

Warn mode issues one warning per kernel. It names the flagged pairs and quotes the first
finding, here an `unverified` line rather than the hazard. `op.alias_check = "strict"` raises
on a potential hazard instead. Each
`with q_free...: pass` is a pure synchronization edge: a channel with no payload is a pair of
barriers.

## 6. The checkers

Tracing a role body checks the local protocol:

- every handle closes exactly once, in the region that opened it;
- producers commit and consumers release;
- every state argument comes back;
- work loops have the drain shape;
- try tokens and previews belong to their epoch;
- recognized TMA copies match the declared bytes.

A violation raises `ProtocolError` at compile time, naming the role, the channel and the
source lines. A notebook cell's lines appear under IPython's hashed file name for the cell.

The TMA check compares the boxes of the copies that complete on `h.barrier` with the
channel's `Tma(bytes)`. If the GEMM declared half its stage bytes, the FULL barrier would
complete when only half the stage had landed.

````python
class BuggyRelay(Relay):
    def __init__(self, bug: str):
        super().__init__()
        self.bug = bug

    @role(warps=0)
    @cute.jit
    def producer(self, ctx, *, ch: Producer):
        lane = cute.arch.lane_idx()
        for e in cutlass.range(ctx.n, unroll=1):
            h = ch.acquire()
            h.data[lane] = e * WORDS + lane
            if cutlass.const_expr(self.bug != "never committed"):
                h.commit()
            if cutlass.const_expr(self.bug == "committed twice"):
                h.commit()
            if cutlass.const_expr(self.bug == "released by the producer"):
                h.release()
        if cutlass.const_expr(self.bug == "end not returned"):
            return None
        return ch


for bug in ("never committed", "committed twice", "released by the producer", "end not returned"):
    try:
        build_relay(BuggyRelay(bug))
    except ProtocolError as error:
        print(f"{bug}:\n  {error}\n")


class HalfBytesGemm(Gemm):
    def __init__(self):
        super().__init__()
        buffers = {"A": self.smem_plan["A"], "B": self.smem_plan["B"]}
        self.ab = Ring(3, Tma((128 + 256) * 64), Umma, buffer=buffers)  # forgot "* 2" for bf16


try:
    compile_tvm_ffi(HalfBytesGemm(), *gemm_fakes(), name="tour_half_bytes")
except ProtocolError as error:
    print(f"half the TMA bytes:\n  {error}")
````

`trace_protocol = True`, set before the first `describe()` or compile, records every traced
acquire/commit/wait/release site. The list holds code sites, not runtime events: a wait in a
loop appears once. Indentation shows block depth (loops and dynamic branches).

````python
traced = Relay(style="token")
traced.trace_protocol = True
build_relay(traced)
show(traced, "Traced")
````

The checks do **not**:

- compare the two sides' epoch counts in dynamic loops;
- prove the dependency graph can make progress;
- validate raw code.

A producer that commits `n + 1` times against a consumer that waits `n` times compiles. So
does a `with` placed under a lane-dependent `if`: predicate the data access, not the handoff.
The [safety contract](guarantees.md) lists what each check means.

## 7. Dropping down

Each escape hatch hands one piece of the protocol back to you, and `describe()` records it.

| Hatch | You take over |
|---|---|
| `h.commit(signal=False)` / `h.release(signal=False)` | one signal and its fences; `h.barrier`, `h.empty_barrier`, `h.index`, `h.phase` give the raw state |
| `Channel(override={"empty_counts": 1, ...})` | a derived count, `tx` or mask |
| `Ring(..., external="consumer")` | a whole side, driven from raw code via `self.barrier_ptr(name)` |
| `Barriers(n, counts=...)` | a reserved mbarrier set with no checks (e.g. for a `cutlass.pipeline`) |
| `@role(..., raw=True)` | a role the layer only places, budgets and dispatches |
| `with unchecked(check, reason=...)` | one trace-time check at one site |

[`attn_gym/warpdrive/examples/raw_interop.py`](https://github.com/meta-pytorch/attention-gym/blob/main/attn_gym/warpdrive/examples/raw_interop.py) mixes them on one ring:

- a managed producer;
- a raw consumer that drives the external side with `cute.arch.mbarrier_wait`/`arrive` and
  its own `PipelineState` (one arrival per stage, declared with `override`);
- a raw watcher that joins a `NamedBarrier` through its native ID.

````python
from attn_gym.warpdrive.examples.raw_interop import (
    RawInterop,
    compile_raw_interop,
    raw_interop,
)

for n in (1, 2, 9):
    wrong, total = raw_interop(n).tolist()
    assert (wrong, total) == (0, sum(range(64 * n))), (n, wrong, total)
print("raw_interop matches for n = 1, 2, 9\n")
interop = RawInterop()
compile_raw_interop(interop)  # describe() reports the named-barrier users after a compile
show(interop, "Roles", "Channels", "Named")
print("\n" + inspect.getsource(RawInterop.consumer))
````

`describe()` marks both raw boundaries:

- the external side's count is the override's (`unverified`);
- `done`'s participation is unverified, because the watcher's arrivals come from a native
  `bar.sync` the layer cannot see.

## Verb cheat sheet

| Verb | Side | Effect |
|---|---|---|
| `h = end.acquire()` / `end.acquire(j)` | producer | wait for an empty stage (slot `j`); arms TMA bytes; advances the end |
| `h.commit()` / `h.retire()` | producer | publish / close a final unused acquire |
| `with end.produce() as h:` | producer | acquire + commit |
| `h = end.wait()` / `end.wait(j)` | consumer | wait for a full stage; advances the end |
| `h.release()` | consumer | hand the stage back (after the side's fences) |
| `with end.consume() as h:` | consumer | wait + release |
| `end.try_wait()` / `try_acquire()` → `Token`; `wait(token=)` / `acquire(token=)` | either | probe once, block only if needed |
| `end.peek()` → `Preview` | consumer | read the next epoch without taking it |
| `end.skip(n)` | exclusive consumer | advance past epochs another role owns |
| `end.views(A=StageView(...))`, `h.view("A", ...)`, `h.data` | either | stage-projected storage |
| `h.barrier`, `h.index`, `h.phase`, `signal=False` | either | raw state for `cute.arch` code |
| `work.valid`, `work.item`, `work.advance()` | any role | drain the work stream |
| `nb.sync()`, `nb.arrive()`, `with seat.turn():` | any role | named-barrier rendezvous, turnstile |
| `op.describe()`, `op.trace_protocol = True`, `op.alias_check = "strict"`, `op.enable_iket()` | host | inspect, trace sites, raise on alias hazards, instrument |
